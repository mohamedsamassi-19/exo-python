```python
import pandas as pd
import seaborn as sns

taxis = sns.load_dataset("taxis")
taxis["pickup"] = pd.to_datetime(taxis["pickup"])
taxis["dropoff"] = pd.to_datetime(taxis["dropoff"])
print(taxis.shape)
print(taxis["pickup"].min())
print(taxis["pickup"].max())
```

```python
taxis["jour"] = taxis["pickup"].dt.date
par_jour = taxis.groupby("jour").agg(
    nb_courses=("total", "count"),
    chiffre_affaires=("total", "sum"),
).round(2)
par_jour
```

```python
taxis["heure"] = taxis["pickup"].dt.hour
courses_par_heure = taxis["heure"].value_counts().sort_index()
heure_plus_chargee = courses_par_heure.idxmax()
print(f"Heure la plus chargée : {heure_plus_chargee} h ({courses_par_heure.max()} courses)")
```

```python
noms_jours = ["lundi", "mardi", "mercredi", "jeudi", "vendredi", "samedi", "dimanche"]
taxis["jour_semaine"] = taxis["pickup"].dt.dayofweek.map(dict(enumerate(noms_jours)))
pourboire_moyen = taxis.groupby("jour_semaine")["tip"].mean().round(2).sort_values(ascending=False)
pourboire_moyen
```

```python
taxis["duree_min"] = (taxis["dropoff"] - taxis["pickup"]).dt.total_seconds() / 60
print(taxis["duree_min"].describe().round(1))

nb_negatives = (taxis["duree_min"] <= 0).sum()
nb_trop_longues = (taxis["duree_min"] > 180).sum()
print("Durées <= 0 :", nb_negatives)
print("Durées > 3 h :", nb_trop_longues)
```

```python
taxis_propres = taxis[(taxis["duree_min"] > 0) & (taxis["duree_min"] <= 180)]
print(len(taxis), len(taxis_propres))
```
