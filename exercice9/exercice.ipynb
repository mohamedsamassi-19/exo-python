```python
import pandas as pd

titanic = pd.read_csv("../exercice8/data/titanic_propre.csv")
```

```python
survie_par_classe_et_sexe = titanic.pivot_table(
    index="class", columns="sex", values="survived", aggfunc="mean"
) * 100
survie_par_classe_et_sexe.round(1)
```

**Réponse 1** : oui, dans chaque classe les femmes ont mieux survécu que les hommes (96,8 % contre 36,9 % en 1re classe, 92,1 % contre 15,7 % en 2e, 50,0 % contre 13,5 % en 3e).

```python
titanic["tranche_age"] = pd.cut(
    titanic["age"], bins=[0, 12, 18, 40, 60, 100],
    labels=["0-12", "13-18", "19-40", "41-60", "60+"]
)
survie_par_tranche_age = titanic.groupby("tranche_age", observed=True)["survived"].mean() * 100
survie_par_tranche_age.round(1)
```
